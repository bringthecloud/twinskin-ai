```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Karena file ini di folder notebooks, kita harus naik satu folder (../) untuk menemukan data
df = pd.read_csv('../data/processed/cosmetic_clean.csv')

print("Data berhasil dibaca!")
print("Jumlah baris & kolom:", df.shape)

# Tampilkan 5 data terakhir
df.head()
```

```python
# Menggunakan Matplotlib untuk membuat grafik

# Hitung jumlah produk per label (kategori)
jumlah_produk = df['Label'].value_counts()

# Buat Grafik Batang
plt.figure(figsize=(10, 6))
plt.bar(jumlah_produk.index, jumlah_produk.values, color='skyblue')

# Menambahkan Judul dan Label
plt.title('Distribusi Kategori Produk', fontsize=15)
plt.xlabel('Nama Kategori (Label)', fontsize=12)
plt.ylabel('Jumlah Produk', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)

# Memutar nama kategori biar gak bertumpuk
plt.xticks(rotation=45)

# Tampilkan grafik
plt.show()
```

```python
# Cari 10 Produk Termahal
produk_mahal = df.sort_values(by='Price', ascending=False).head(10)

# Buat Grafik Batang Horizontal (biar nama produknya kebaca)
plt.figure(figsize=(12, 6))
plt.barh(produk_mahal['Name'], produk_mahal['Price'], color='gold')

plt.title('10 Produk Kosmetik Termahal', fontsize=15)
plt.xlabel('Harga', fontsize=12)
plt.ylabel('Nama Produk', fontsize=12)
plt.grid(True, axis='x', linestyle='--', alpha=0.7)

plt.show()

# Cek siapa pemilik brand termahal
print("10 Produk Termahal:")
print(produk_mahal[['Name', 'Brand', 'Price']])
```

```python
# Analisa Produk berdasarkan Jenis Kulit
# Kolom target: Combination, Dry, Normal, Oily, Sensitive
kolom_kulit = ['Combination', 'Dry', 'Normal', 'Oily', 'Sensitive']

# Hitung jumlah produk yang cocok untuk tiap jenis kulit
# (Angka 1 berarti cocok, kita jumlahkan semua angka 1 itu)
jumlah_kulit = df[kolom_kulit].sum().sort_values(ascending=False)

print(jumlah_kulit)

# Buat Grafik
plt.figure(figsize=(10, 6))
plt.bar(jumlah_kulit.index, jumlah_kulit.values, color='salmon')

plt.title('Jumlah Produk berdasarkan Tipe Kulit', fontsize=15)
plt.xlabel('Jenis Kulit', fontsize=12)
plt.ylabel('Jumlah Produk', fontsize=12)
plt.grid(True, axis='y', linestyle='--', alpha=0.7)

plt.show()
```
